```python
from music21 import * # import everything from music21
```

```python
s = stream.Stream() # create a stream object
s.append(meter.TimeSignature('4/4')) # append a time signature
s.append(note.Note('C4', quarterLength = 0.5)) 
s.append(note.Note('D4', quarterLength = 0.5))
s.append(note.Rest(quarterLength = 1))
s.append(note.Note('E4', quarterLength = 2.5))
s.append(note.Note('F4', quarterLength = 1))
s.append(note.Note('G4', quarterLength = 1))
```

```python
s.show() # show the score
```

```python
s.show('midi') # play the score
```
