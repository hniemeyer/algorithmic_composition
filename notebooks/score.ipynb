```python
from music21 import *
```

```python
my_score = stream.Score()
part1 = stream.Part()
part2 = stream.Part()
```

```python
voice1 = stream.Voice()
voice2 = stream.Voice()
voice3 = stream.Voice()
measure = stream.Measure()
measure.append(meter.TimeSignature('4/4'))
measure2 = stream.Measure()
measure2.append(clef.BassClef())
measure2.append(meter.TimeSignature('4/4'))
```

```python
notes = ["C4", "D4", "E4", "F4"]
for note_name in notes:
    melody_note = note.Note(note_name)
    voice1.append(melody_note)
    voice2.append(melody_note.transpose("M3"))
    voice3.append(melody_note.transpose("P5"))
    measure2.append(melody_note.transpose(-12))
```

```python
measure.insert(0, voice1)
measure.insert(0, voice2)
measure.insert(0, voice3)
part1.append(measure)
part2.append(measure2)
```

```python
my_score.insert(0, part1)
my_score.insert(0,part2)
```

```python
my_score.show()
```
