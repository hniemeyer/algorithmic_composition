```python
from music21 import  *
from arvo import isorhythm
from arvo import minimalism
from arvo import scales
from arvo import transformations
import copy
```

```python
pitches_a = ["C5", "D5", "E5"]
durations_a = [1.0, 2.0, 0.5, 0.25]
```

```python
res_a = isorhythm.create_isorhythm(pitches_a, durations_a)
```

```python
res_a.show()
```

```python
res_a.show("midi")
```

```python
pattern_a = minimalism.additive_process(res_a)
```

```python
pattern_a.show()
```

```python
pattern_a.show("midi")
```

```python
pattern_a
```

```python
pitches_b = ["C3", "D3", "E3"]
durations_b = [1.0, 2.0, 0.5, 0.25]
res_b = isorhythm.create_isorhythm(pitches_b, durations_b)
```

```python
pattern_b = minimalism.subtractive_process(res_b)
```

```python
pattern_b.show()
```

```python
pattern_b.show("midi")
```
