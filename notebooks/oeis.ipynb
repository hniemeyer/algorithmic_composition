```python
available_pitches = ["C","D","E","F","G","A","B"]
print(f"number of pitches: {len(available_pitches)}")
available_chord_types = ["maj", "min", "aug", "dim", "maj7", "min7", "7", "Maj9", "min9", "9"]
print(f"number of chord types: {len(available_chord_types)}")
available_durations = [1.0,2.0,4.0, 0.5, 0.25]
my_stream = stream.Stream()
for oeis_seq_element in oeis_sequence:
    current_pitch = available_pitches[oeis_seq_element % len(available_pitches)]
    current_chord_type = available_chord_types[oeis_seq_element % len(available_chord_types)]
    current_duration = available_durations[oeis_seq_element % len(available_durations)]
    current_chord_name = current_pitch + current_chord_type
    print(current_chord_name)
    print(current_duration)
    current_chord_symbol = harmony.ChordSymbol(current_chord_name)
    current_chord = chord.Chord(current_chord_symbol.pitches, quarterLength=current_duration).transpose(12)
    my_stream.append(current_chord)
```

```python
my_stream.show()
```

```python
my_stream.show("midi")
```

```python
my_stream.write("midi", "primes.midi")
```
